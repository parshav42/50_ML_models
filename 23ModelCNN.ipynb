🧠 Concept First

CNN works like this:

Normal NN → flattens image → loses spatial info
CNN → scans image in small windows → keeps spatial info
Like human eye scanning a photo piece by piece
Conv layer → finds edges, shapes, patterns
MaxPool layer → shrinks image, keeps important parts
Stack them → finds complex patterns!

```python
import torch
import torchvision.datasets
import torchvision.transforms
from torchvision import datasets
from torchvision.transforms import ToTensor
```

```python
train_data = datasets.FashionMNIST(
    root = '/content/data',
    train = True,
    download = True,
    transform=ToTensor(),
    target_transform = None
)

test_data = datasets.FashionMNIST(
    root = '/content/data',
    train = False,
    download = True,
    transform=ToTensor(),
)
```

```python
from torch.utils.data import DataLoader

train_dataloder = DataLoader(
    train_data,
    batch_size=64,
    shuffle=True

    )

test_dataloader = DataLoader(
    test_data,
    batch_size=64,
    shuffle=False
)
```

```python
print(len(train_dataloder))
```

```python
from torch import nn

class mo(nn.Module):
    def __init__(self,input):
        super().__init__()

        self.mo = nn.Sequential(

            # Block 1 — First Conv
            nn.Conv2d(1, 32, kernel_size=3),
            nn.ReLU(),
            nn.MaxPool2d(2),

            # Block 2 — Second Conv
            nn.Conv2d(32, 64, kernel_size=3),
            nn.ReLU(),
            nn.MaxPool2d(2),

            # Flatten
            nn.Flatten(),

            # Fully Connected
            nn.Linear(64 * 5 * 5, 128),
            nn.ReLU(),
            nn.Linear(128, 10)
        )
    def forward(self, x):
        return self.mo(x)
```

```python
model = mo(1)
loss_fn = nn.CrossEntropyLoss()

optim = torch.optim.Adam(model.parameters(),lr=0.01)
```

```python
epos = 5
for i in range(epos):
    # Iterate over the DataLoader, not the raw dataset
    for batch, (X, y) in enumerate(train_dataloder):
        model.train()
        y_pre = model(X)

        loss = loss_fn(y_pre, y)

        optim.zero_grad()

        loss.backward()

        optim.step()

    print(f"Epoch {i+1} completed.")
```

```python
model.eval()
test_loss, correct = 0, 0

with torch.no_grad():
    for X, y in test_dataloader:
        # Forward pass
        pred = model(X)
        test_loss += loss_fn(pred, y).item()
        correct += (pred.argmax(1) == y).type(torch.float).sum().item()

test_loss /= len(test_dataloader)
correct /= len(test_dataloader.dataset)

print(f"Test Error: \n Accuracy: {(100*correct):>0.1f}%, Avg loss: {test_loss:>8f} \n")
```
