PCA is a dimensionality reduction technique:

Imagine you have 50 features in your dataset
Too many features → slow model, confusing data
PCA compresses them into 2 or 3 features
But keeps maximum information
Like squeezing a big file into a zip — less size same content
Great for visualization and speeding up models

Real example:

50 features → PCA → 2 features
Now you can plot on a 2D graph!
See patterns impossible to see before

 Download Dataset
Go to Kaggle
Search "Breast Cancer Wisconsin"
Has 30 features — perfect for PCA!
Target → Malignant or Benign

 What To Build

Model Part 1 — Reduce Dimensions:

Apply PCA with n_components=2
Transform your 30 features into just 2
Print how much variance explained by each component
Add both percentages — total variance kept?

Model Part 2 — Visualize:

Scatter plot of 2 PCA components
Color by target → Malignant vs Benign
Can you see clear separation between two groups?

Model Part 3 — Find Best Components:

Try n_components = 2, 5, 10, 15, 20
Plot explained variance ratio for each
Find how many components keep 95% of information

Model Part 4 — Compare Speed:

Train Logistic Regression on original 30 features → check accuracy
Train Logistic Regression on PCA 2 features → check accuracy
Which is faster? Which is more accurate?
