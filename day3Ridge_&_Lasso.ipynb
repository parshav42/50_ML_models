```python
import numpy as np
import pandas as pp
import torch
from sklearn.model_selection import train_test_split
```

```python
hp = pp.read_csv('/content/housepricecorrect.csv')
hp.head()
```

```python
X = hp.iloc[:,0:7]
y = hp.iloc[:,-1]
```

```python
X_train ,X_test , y_train , y_test = train_test_split(X,y,test_size=0.2)
```

```python
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()

y_test = sc.fit_transform(y_test.values.reshape(-1, 1))
y_train = sc.fit_transform(y_train.values.reshape(-1, 1))
```

```python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()

lr.fit(X_train , y_train)
```

```python
y_pre = lr.predict(X_test)
```

```python
from sklearn.metrics import r2_score


r = r2_score(y_test,y_pre)
```

```python
r
```

```python
from sklearn.linear_model import RidgeCV
ridge_cv = RidgeCV(alphas=[0.1, 1.0, 10.0,100.0], cv=5)
ridge_cv.fit(X_train, y_train)
```

```python
y_prep2 = ridge_cv.predict(X_test)
```

```python
from sklearn.metrics import r2_score

r = r2_score(y_test,y_prep2)
```

```python
r
```

```python
from sklearn import linear_model
clf = linear_model.Lasso(alpha=0.001)

clf.fit(X_train,y_train)

```

```python
y_pre = clf.predict(X_test)
```

```python
from sklearn.metrics import r2_score


r = r2_score(y_test,y_pre)
```

```python
r
```
