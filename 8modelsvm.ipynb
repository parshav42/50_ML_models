SVM works like this:

Imagine two groups of points on a paper
SVM draws the best possible line between them
That line is called hyperplane
It tries to keep maximum gap between both groups
That gap is called margin
Points closest to line are called support vectors

```python
X = tt.drop('Loan_Status',axis=1)
y=tt['Loan_Status']
```

```python
from sklearn.preprocessing import  StandardScaler

sc =  StandardScaler()

t = sc.fit_transform(X)
```

```python
X = pp.DataFrame(t,columns=sc.get_feature_names_out())
# X = pp.concat()
```
