```python
ak = em['BuildingArea'].mode().values[0]
em['BuildingArea'] = em['BuildingArea'].replace(np.nan,ak)
```

```python
sk = em['YearBuilt'].mode().values[0]
em['YearBuilt'] = em['YearBuilt'].replace(np.nan,sk)

```

```python
ak = em['CouncilArea'].mode().values[0]
em['CouncilArea'] = em['CouncilArea'].replace(np.nan,ak)
```

```python
ak = em['Car'].min()
em['Car'] = em['Car'].replace(np.nan,ak)
```

```python
em.drop(['Lattitude'],axis =1,inplace=True)
```

```python
em['Date'] = pp.to_datetime(em['Date'],dayfirst=True)
```
