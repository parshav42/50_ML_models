Transfer Learning works like this:

Training CNN from scratch → needs millions of images + days
Transfer Learning → use a model already trained on millions of images
Just teach it your specific task at the end
Like hiring an expert instead of training from zero!

ResNet18:

Trained on ImageNet → 1.2 million images, 1000 classes
Already knows edges, shapes, textures, objects
You just replace the last layer for your task!

📌 What To Build

Step 1 — Load Pretrained ResNet18:

Import from torchvision.models
Load resnet18(pretrained=True)
Print the model — see its architecture!

Step 2 — Freeze All Layers:

Loop through model.parameters()
Set param.requires_grad = False
This freezes pretrained weights — don't change them!

Step 3 — Replace Last Layer:

ResNet last layer → model.fc
Replace with nn.Linear(512, 10) for CIFAR-10
Only this layer will be trained!

Step 4 — Transform Images:

ResNet needs 224×224 images
CIFAR is 32×32 — resize it!
Add transforms.Resize(224) to your transform

Step 5 — Train:

Loss → CrossEntropyLoss
Optimizer → Adam lr=0.001
Only train last layer parameters!
Epochs → 5
📌 Two Approaches — Compare Both!

Approach A — Feature Extraction:

Freeze ALL pretrained layers
Only train last layer
Fast training!

Approach B — Fine Tuning:

Unfreeze ALL layers
Train everything with small lr=0.0001
Slower but better accuracy!

```python
from torchvision import transforms

t = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize(
        mean = [0.5,0.5,0.5],
        std = [0.5,0.5,0.5]
    )
])
```

```python
from torchvision import datasets



trin_t =  datasets.CIFAR10(
    root = '/content/data',
    train = True,
    download = True,
    transform = t,
    target_transform = None

)

test_t = datasets.CIFAR10(
    root = '/content/data',
    train = False,
    download = True,
    transform = t
)
```

```python
from torch.utils.data import DataLoader


train = DataLoader(
    trin_t,
    batch_size = 32,
    shuffle = True,
    num_workers = 2
)
test = DataLoader(
    test_t,
    batch_size = 32,
    num_workers= 2
)
```
