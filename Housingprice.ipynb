```python
X = hp.iloc[:,1:7]
y = hp.iloc[:,0]
```

```python
X
```

```python
from sklearn.preprocessing import OneHotEncoder

select_feature = X.select_dtypes(include=['object']).columns

oh = OneHotEncoder(sparse_output=False,drop='first')

after_enco = oh.fit_transform(X[select_feature])

enco_f = pp.DataFrame(after_enco,columns=oh.get_feature_names_out(select_feature))

final_df = pp.concat([X.drop(columns=select_feature),enco_f],axis=1)
```

```python
# from sklearn.preprocessing import StandardScaler


# scalr = StandardScaler()

# final_df_scalar=scalr.fit_transform(final_df)

```

```python
# final_df_new = pp.DataFrame(final_df_scalar,columns=final_df.columns)
```

```python
X_train,X_test,y_train,y_test = train_test_split(final_df,y,test_size =0.2)
```

```python
X_train.shape,X_test.shape,y_train.shape,y_test.shape
```

```python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()


lr.fit(X_train,y_train)
```

```python
y_prep = lr.predict(X_test)
```

```python

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

print('mae',mean_absolute_error(y_test,y_prep))

print('mse',mean_squared_error(y_test,y_prep))
print('r2',r2_score(y_test,y_prep))
```
