```python
from difflib import get_close_matches

correct_values = ['Pigmentation', 'Dryness', 'Acne']

def clean_gender(x):
    x = str(x).strip().lower()
    match = get_close_matches(x, correct_values, n=1, cutoff=0.75)
    return match[0] if match else "Unknown"

sk["SkinConcerns_clean"] = sk["SkinConcerns"].apply(clean_gender)
```
