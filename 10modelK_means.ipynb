K-Means is completely different from all previous models:

All previous models → Supervised (you gave labels)
K-Means → Unsupervised (no labels — model finds groups itself)
Example → Give customer data → model finds groups automatically
Group 1 → High income, low spending
Group 2 → Low income, high spending
Group 3 → High income, high spending
You decide how many groups → that is K
