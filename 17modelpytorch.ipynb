In Sklearn → model.fit() does everything automatically
In PyTorch → you build everything yourself:

Define the model
Define loss function
Define optimizer
Write training loop manually
This gives you full control!
📥 No Dataset Needed
Create fake data yourself using torch
Make 100 points with a linear relationship
Add some random noise to make it realistic
📌 What To Build

Step 1 — Create Data:

X = 100 random numbers
y = 3 * X + 7 + some noise
So true weight = 3, true bias = 7
Can your model learn these values?

Step 2 — Build Model:

Create a class called LinearModel
Inherit from nn.Module
Add one nn.Linear layer
Write forward function

Step 3 — Setup Training:

Loss function → nn.MSELoss()
Optimizer → torch.optim.SGD
Learning rate → start with 0.01

Step 4 — Training Loop:

Loop for 100 epochs
Each epoch:
Forward pass → get predictions
Calculate loss
Zero gradients → optimizer.zero_grad()
Backward pass → loss.backward()
Update weights → optimizer.step()
Print loss every 10 epochs

Step 5 — Check Results:

Print learned weight — is it close to 3?
Print learned bias — is it close to 7?
Plot original data vs model predictions
🧠 Understand This
Why do we call optimizer.zero_grad() every epoch?
What is forward pass?
What is backward pass?
What is epoch?
What does loss going down mean?



Target
Loss should decrease every epoch
Final weight close to 3.0
Final bias close to 7.0
Plot showing good fit line!

```python
import torch

```

```python
if torch.cuda.is_available():
    device = torch.device('cuda')
else:
    device = torch.device('cpu')
```

```python
X = torch.rand([1,100])
y = 3 *X + 7 + torch.rand(X.shape)*0.01
```

```python
X
```

```python
from torch import nn

class linea(nn.Module):
    def __init__(self):
        super(). __init__()

        self.li = nn.Linear(1,1)

    def forward(self,X):
        x = self.li(X)
        return x
```

```python
model = linea()
loss = nn.MSELoss()
optim = torch.optim.SGD(model.parameters(),lr=0.01)
```
