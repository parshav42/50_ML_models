DBSCAN is different from K-Means:

K-Means → you tell how many clusters
DBSCAN → finds clusters automatically
Works by finding dense regions of points
Points in empty areas → marked as noise/outliers
Great for finding odd shaped clusters
Real use → finding fraud transactions, anomalies

Two important words:

eps → how close points must be to be neighbors
min_samples → minimum points to form a cluster
Dataset
Use same Mall Customer dataset
Or search Kaggle → "Credit Card Dataset for Clustering"
Download new one for more practice

Data Prep:

Load dataset

Scale using StandardScaler

Model:

Train DBSCAN with default settings first
Print how many clusters it found
Print how many points marked as noise (-1)
Then tune:
eps → try 0.1, 0.3, 0.5, 1.0
min_samples → try 3, 5, 10
See how clusters change with each value

Visualize:

Color each cluster differently

```python
import torch
import numpy as np
import pandas as pp
import seaborn as ss
from sklearn.model_selection import train_test_split
```

```python
mp = pp.read_csv("/content/Mall_Customers.csv")
```

```python
mp.head()
```

```python
X= mp.iloc[:,3:5]
```

```python
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()

x =sc.fit_transform(X)
```

```python
X_final = pp.DataFrame(x,columns=sc.get_feature_names_out())
X_final
```

```python
from sklearn.cluster import DBSCAN

s = DBSCAN(eps=0.2,min_samples=5)

clusters = s.fit(X_final)
```

```python
s.labels_
```

```python
X_final.iloc[:, 0]
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 6))
scatter = plt.scatter(X_final.iloc[:, 0], X_final.iloc[:, 1], c=s.labels_, cmap='viridis')
plt.colorbar(scatter)
plt.title('DBSCAN Clustering Results')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()
```
