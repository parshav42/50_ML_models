This is not a new model — this is a better way to build models!

Pipeline:

Combines all steps into one clean flow
Step 1 → Scale data
Step 2 → Train model
All in one single object
No chance of data leakage!

Cross Validation:

Normal split → train on 80%, test on 20%
CV → splits data into 5 parts
Trains 5 times, tests on different part each time
Final score = average of all 5 scores
Much more reliable than single split!
What To Build

Part 2 — Cross Validation:

Use cross_val_score from sklearn
