```python
import torch
import pandas as pp
import numpy as np
import sklearn as sn
import seaborn as sn
```

```python
sp = pp.read_csv('/content/spam.csv')
```

```python
sp['Category'].value_counts()
```

```python
sp.isnull().sum()
```

```python
X = sp.drop('Category',axis=1)
y=sp['Category']
```

```python
from sklearn.feature_extraction.text import CountVectorizer

cr = CountVectorizer()

X_tr = cr.fit_transform(X['Message'])
```

```python
print("Vocabulary:", cr.vocabulary_)
```

```python
cr.get_feature_names_out()[400:500]
```

```python
from sklearn.preprocessing import LabelEncoder

lr = LabelEncoder()

y_tar=lr.fit_transform(y)
```

```python
X_tr.
```

```python
from sklearn.model_selection  import train_test_split
X_train , X_test,y_train,y_test = train_test_split(X_tr,y_tar ,test_size=0.25,random_state=42)
```

```python
from sklearn.naive_bayes import MultinomialNB

m_nb = MultinomialNB()
m_nb.fit(X_train, y_train)
```

```python
y_pre = m_nb.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, recall_score, precision_score, f1_score

print("Accuracy score:", accuracy_score(y_test, y_pre))
print("Classification report:\n", classification_report(y_test, y_pre))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pre))
print("Recall Score:\n", recall_score(y_test, y_pre))
print("Precision score:\n", precision_score(y_test, y_pre))
print("F1 score:\n", f1_score(y_test, y_pre)
```
