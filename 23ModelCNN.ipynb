🧠 Concept First

CNN works like this:

Normal NN → flattens image → loses spatial info
CNN → scans image in small windows → keeps spatial info
Like human eye scanning a photo piece by piece
Conv layer → finds edges, shapes, patterns
MaxPool layer → shrinks image, keeps important parts
Stack them → finds complex patterns!


📥 Download MNIST — No Kaggle Needed!

📌 Architecture To Build
Input → (1, 28, 28) grayscale image
    ↓
Conv2d(1, 32, kernel_size=3) + ReLU
MaxPool2d(2)
    ↓
Conv2d(32, 64, kernel_size=3) + ReLU
MaxPool2d(2)
    ↓
Flatten()
Linear(64 * 5 * 5, 128) + ReLU
Linear(128, 10)
